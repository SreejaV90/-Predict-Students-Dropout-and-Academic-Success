# Project Title
Predict Students' Dropout and Academic Success

## Domain
Education Analytics and Machine Learning

## Project Overview

Education plays an important role in shaping an individual's future. However, student dropout has become a significant challenge for educational institutions worldwide. Identifying students who may be at risk of dropping out and understanding the factors associated with academic success can help institutions improve student support and academic planning.

This project focuses on analyzing student-related data to understand the patterns and characteristics associated with academic outcomes. Using the "Predict Students' Dropout and Academic Success" dataset from the UCI Machine Learning Repository, the project examines 4,424 student records containing 36 input features.

The dataset includes information related to students' demographic backgrounds, admission details, previous qualifications, socioeconomic conditions, academic performance, and other relevant indicators.

The primary objective is to analyze these factors and develop a machine learning classification model that predicts student outcomes into three categories: Dropout, Enrolled, and Graduate.

## Business Problem

Educational institutions often face difficulties in identifying students who are at risk of discontinuing their studies. Although institutions collect large amounts of student information, extracting meaningful insights from these records requires systematic data analysis.

Without proper analysis, it becomes difficult to understand how demographic characteristics, admission-related factors, socioeconomic conditions, and academic performance are associated with student outcomes.

Another challenge is identifying the academic indicators that distinguish students who successfully graduate from those who drop out or remain enrolled.

This project addresses these challenges through data cleaning, exploratory data analysis, statistical investigation, and machine learning. It aims to identify meaningful patterns in student data and develop a classification model to predict academic outcomes.

## Project Objectives

1. **Understanding Student Outcomes:** Analyze the distribution of students across the three target categories: Dropout, Enrolled, and Graduate.

2. **Demographic and Admission Analysis:** Explore demographic characteristics, admission information, previous qualifications, and other student-related attributes in relation to academic outcomes.

3. **Academic Performance Analysis:** Investigate first- and second-semester academic performance, including grades, curricular units enrolled, evaluated, and approved.

4. **Socioeconomic Analysis:** Examine the relationship between available socioeconomic indicators, such as scholarship status and tuition-fee information, and student outcomes.

5. **Identifying Important Factors:** Explore the variables that show meaningful associations with student dropout and academic success.

6. **Data Quality and Preprocessing:** Identify and review missing values, duplicate records, unusual values, and inconsistent data formats. Prepare the dataset for reliable analysis.

7. **Model Evaluation:** Compare model performance using accuracy, precision, recall, F1-score, macro F1-score, and confusion matrix.

8. **Data Visualization and Insights:** Create meaningful charts and visualizations to communicate student performance patterns and analytical findings.

## Key Feature Groups

The dataset contains several categories of student-related information:

- **Demographic Information:** Student demographic characteristics and background information.

- **Admission Information:** Application mode, application order, course selection, admission-related details, and previous qualification information.

- **Socioeconomic Information:** Variables related to family background, scholarship status, tuition fees, and other socioeconomic indicators.

- **Academic Performance:** First- and second-semester academic records, including curricular units enrolled, evaluated, approved, and semester grades.

- **Macroeconomic Indicators:** Selected economic indicators that provide additional context for the analysis.

- **Target:** The final academic outcome of the student, classified as Dropout, Enrolled, or Graduate.

The official UCI dataset documentation will be used to understand the definitions and encoded values of individual features.
